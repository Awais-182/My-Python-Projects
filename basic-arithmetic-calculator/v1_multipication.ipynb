## 📌 Overview
This notebook contains **Version 3** of the Basic Arithmetic Calculator. It prompts the user for two numbers, performs multiplication, and displays the result clearly.

### 🌟 Key Features:
* Accepts dynamic user input using `input()`.
* Performs basic multiplication (`num1 * num2`).
* Uses formatted print output to display the calculation clearly.

---


```python
first_num=input("Enter The First Number: ")
second_num=input("Enter The Second Number: ")
result=int(first_num)*int(second_num)
print("Your result is:",result)
```


## ⚠️ Known Limitations & Design Decisions
* **Integer Logic:** Currently uses `int()`, which means decimal inputs will throw a `ValueError`.
* **Single Execution:** Runs once and terminates (loops will be added in upcoming versions).
