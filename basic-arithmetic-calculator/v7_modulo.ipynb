## 📌 Program Description
This program is the sixth entry in the **Basic Arithmetic Calculator** series within the `python-program-vault`.

### 🎯 Objectives:
1. Accept two numeric inputs from the user via the console (a **dividend** and a **divisor**).
2. Convert user input into Python integer variables.
3. Perform a modulo operation (`%`) to calculate the remainder after integer division.
4. Output the resulting remainder formatted clearly to the screen.

---
## 📥 Step 1: Input Acquisition

### Explanation:
* We use the `input()` function to prompt the user for raw text input.
* We convert the string values directly into integers using `int()`.
* The first input represents the **dividend** (number to be divided), and the second input represents the **divisor** (number doing the dividing).

### Code Snippet:
```python
num1 = int(input("Enter dividend (first number): "))
num2 = int(input("Enter divisor (second number): "))

---
## ⚙️ Step 2: Modulo Execution

### Explanation:
* We use Python's built-in modulo operator (`%`).
* The expression `num1 % num2` divides `num1` by `num2` and yields only the integer remainder left over (e.g., `17 % 5` yields `2` because 5 fits into 17 three times with a remainder of 2).

### Code Snippet:
```python
result = num1 % num2

---
## 📤 Step 3: Result Display

### Explanation:
* We print the formatted output using simple comma-separated concatenation inside `print()`.

### Code Snippet:
```python
print("The remainder of", num1, "%", num2, "is:", result)

````python
first_num=int(input("Enter The First Number: "))
second_num=int(input("Enter The Second Number: "))
result=first_num%second_num
print("Your result is:",result)
````
