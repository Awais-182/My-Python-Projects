## 📌 Program Description
This program is the seventh entry in the **Basic Arithmetic Calculator** series within the `python-program-vault`.

### 🎯 Objectives:
1. Accept two numeric inputs from the user via the console (a **base** and an **exponent**).
2. Convert user input into Python integer variables.
3. Perform an exponentiation operation (`**`) to calculate the base raised to the power of the exponent.
4. Output the resulting power formatted clearly to the screen.

---
## 📥 Step 1: Input Acquisition

### Explanation:
* We use the `input()` function to collect raw values from the console.
* We convert those values to integers using `int()`.
* The first value represents the **base number**, and the second value represents the **power (exponent)** to which the base will be raised.

### Code Snippet:
```python
base = int(input("Enter base number: "))
exponent = int(input("Enter exponent (power): "))

---
## ⚙️ Step 2: Exponentiation Execution

### Explanation:
* We use Python's built-in exponentiation operator (`**`).
* The expression `base ** exponent` multiplies the `base` by itself as many times as specified by `exponent` (e.g., `2 ** 3` computes $2 \times 2 \times 2 = 8$).

### Code Snippet:
```python
result = base ** exponent

---
## 📤 Step 3: Result Display

### Explanation:
* We print the formatted output using simple comma-separated concatenation inside `print()`.

### Code Snippet:
```python
print("The result of", base, "raised to the power of", exponent, "is:", result)

````python
first_num=int(input("Enter The First Number: "))
second_num=int(input("Enter The Second Number: "))
result=first_num**second_num
print("Your result is:",result)
````
