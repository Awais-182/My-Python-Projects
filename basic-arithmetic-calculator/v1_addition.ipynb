### 📌 Overview
This notebook contains **Version 1** of the Basic Arithmetic Calculator program. It takes two user-defined numbers, converts them to numerical data types (`float`), performs addition, and displays the result clearly.

#### Key Features:
* Accepts dynamic user input using `input()`.
* Handles both whole numbers (integers) and decimals (`float`).
* Uses formatted print output.

```python
first_num=int(input("Enter The First Number: "))
second_num=int(input("Enter The Second Number: "))
result=first_num+second_num
print("Your result is:",result)
```

---

### ⚠️ Program Limitations & Known Edge Cases

This initial implementation has several key limitations that are tracked for future versions:

1. **Integer-Only Data Type (`int()` constraint):**
   * Inputs are cast directly using `int()`.
   * **Failure Case:** Entering decimal values (e.g., `5.5`) causes Python to raise a `ValueError` and crash execution.

2. **No Input Validation:**
   * The program assumes valid numeric input without exception handling (`try-except`).
   * **Failure Case:** Entering letters, special characters, or leaving the prompt blank results in an unhandled exception crash.

3. **Single-Run Execution:**
   * The program executes once and terminates immediately without allowing repeated calculations.
