## 📌 Program Description
This program is the fourth entry in the **Basic Arithmetic Calculator** series within the `python-program-vault`.

### 🎯 Objectives:
1. Accept two numeric inputs from the user via the console.
2. Convert user input into Python integer variables.
3. Perform a division operation (`/`) on the stored numbers.
4. Output the exact quotient formatted clearly to the screen.

---
## 📥 Step 1: Input Acquisition

### Explanation:
* We use the built-in `input()` function to prompt the user for input.
* Since `input()` returns string data, we wrap it in `int()` to convert the characters into mathematical integers.
* The first number represents the **dividend** (numerator) and the second represents the **divisor** (denominator).

### Code Snippet:
```python
num1 = int(input("Enter dividend (first number): "))
num2 = int(input("Enter divisor (second number): "))

---
## ⚙️ Step 2: Division Execution

### Explanation:
* We apply the standard forward slash operator (`/`) to perform float division.
* **Note on Behavior:** In Python 3, using `/` automatically converts the result to a floating-point number (`float`), even if both inputs are integers and divide evenly (e.g., `10 / 2` evaluates to `5.0`).

### Code Snippet:
```python
result = num1 / num2

---
## 📤 Step 3: Result Display

### Explanation:
* We use standard string concatenation with commas in the `print()` statement to combine static text labels and numerical variables (`num1`, `num2`, `result`).

### Code Snippet:
```python
print("The result of", num1, "/", num2, "is:", result)

````python
first_num=int(input("Enter The First Number: "))
second_num=int(input("Enter The Second Number: "))
result=first_num/second_num
print("Your result is:",result)
````
